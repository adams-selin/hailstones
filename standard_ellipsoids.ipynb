# Code to generate reference 3D ellipsoid STL files for printing/measurements

```python
import numpy as np
from matplotlib import cm
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import pyvista as pv
import trimesh
from copy import deepcopy
import pymeshfix as mf
%matplotlib widget
```

## Now generate different ellipsoids

```python
#Use pyvista to generate an oblate spheroid. Increase u_res, v_res, w_res will increase file size.
#https://docs.pyvista.org/api/utilities/_autosummary/pyvista.parametricellipsoid
ellipsoid_surf_x3y5z2 = pv.ParametricEllipsoid(xradius = 30, yradius=50, zradius=20)
ellipsoid_surf_x3y3z2 = pv.ParametricEllipsoid(xradius = 30, yradius=30, zradius=20)
ellipsoid_surf_x3y3z3 = pv.ParametricEllipsoid(xradius = 30, yradius=30, zradius=30)

ellipsoid_surf_x2y4z1 = pv.ParametricEllipsoid(xradius = 20, yradius=40, zradius=10)
ellipsoid_surf_x2y2z1 = pv.ParametricEllipsoid(xradius = 20, yradius=20, zradius=10)


#triangulate
ellipsoid_surf_x3y5z2 = ellipsoid_surf_x3y5z2.triangulate()
ellipsoid_surf_x3y3z2 = ellipsoid_surf_x3y3z2.triangulate()
ellipsoid_surf_x3y3z3 = ellipsoid_surf_x3y3z3.triangulate()

ellipsoid_surf_x2y4z1 = ellipsoid_surf_x2y4z1.triangulate()
ellipsoid_surf_x2y2z1 = ellipsoid_surf_x2y2z1.triangulate()

```

## Plot the results

```python
pl = pv.Plotter(shape=(1, 2))
pl.subplot(0, 0)
_ = pl.add_mesh(ellipsoid_surf_x3y5z2,color='blue')
pl.subplot(0, 1)
_ = pl.add_mesh(ellipsoid_surf_x2y2z1 ,color='blue')
pl.show()
```

## Write stuff out

```python
# Convert pyvista polydata to trimesh and export
ellipsoid_x3y5z2_tm = trimesh.Trimesh(vertices=ellipsoid_surf_x3y5z2.points, faces=ellipsoid_surf_x3y5z2.faces.reshape(-1, 4)[:, 1:])
ellipsoid_x3y3z2_tm = trimesh.Trimesh(vertices=ellipsoid_surf_x3y3z2.points, faces=ellipsoid_surf_x3y3z2.faces.reshape(-1, 4)[:, 1:])
ellipsoid_x3y3z3_tm = trimesh.Trimesh(vertices=ellipsoid_surf_x3y3z3.points, faces=ellipsoid_surf_x3y3z3.faces.reshape(-1, 4)[:, 1:])
ellipsoid_x2y4z1_tm = trimesh.Trimesh(vertices=ellipsoid_surf_x2y4z1.points, faces=ellipsoid_surf_x2y4z1.faces.reshape(-1, 4)[:, 1:])
ellipsoid_x2y2z1_tm = trimesh.Trimesh(vertices=ellipsoid_surf_x2y2z1.points, faces=ellipsoid_surf_x2y2z1.faces.reshape(-1, 4)[:, 1:])

# Export to STL files
ellipsoid_x3y5z2_tm.export('ellipsoid_surf_x3y5z2.stl');
ellipsoid_x3y3z2_tm.export('ellipsoid_surf_x3y3z2.stl');
ellipsoid_x3y3z3_tm.export('ellipsoid_surf_x3y3z3.stl');

ellipsoid_x2y4z1_tm.export('ellipsoid_surf_x2y4z1.stl');
ellipsoid_x2y2z1_tm.export('ellipsoid_surf_x2y2z1.stl');
```
